```python
# Overall attrition KPI
```

```python
total_employees = len(df)
total_exited = df["Attrition"].sum()
total_retained = total_employees - total_exited

attrition_rate = (total_exited / total_employees) * 100

print("Total Employees:", total_employees)
print("Employees Exited:", total_exited)
print("Employees Retained:", total_retained)
print("Overall Attrition Rate: {:.2f}%".format(attrition_rate))
```

```python
# Create KPI table
```

```python
kpi = pd.DataFrame({
    "Metric": [
        "Total Employees",
        "Employees Exited",
        "Employees Retained",
        "Overall Attrition Rate"
    ],
    "Value": [
        total_employees,
        total_exited,
        total_retained,
        f"{attrition_rate:.2f}%"
    ]
})

kpi
```

```python
# Department wise attrition
```

```python
department_attrition = (
    df.groupby("Department")["Attrition"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

department_attrition.columns = [
    "Department",
    "Employees",
    "Exited",
    "Attrition_Rate"
]

department_attrition["Attrition_Rate"] *= 100

department_attrition
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=department_attrition,
    x="Department",
    y="Attrition_Rate"
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")

plt.show()
```

```python
hotspots = pd.DataFrame({
    "Factor": [
        "Overall",
        "OverTime = Yes",
        "BusinessTravel = Travel Frequently",
        "Age <=25",
        "Tenure = 0-2 years",
        "JobRole = Sales Representative",
        "JobRole = Laboratory Technician"
    ],
    "Attrition_Rate": [
        attrition_rate,
        df.loc[df["OverTime"] == "Yes", "Attrition"].mean() * 100,
        df.loc[df["BusinessTravel"] == "Travel_Frequently", "Attrition"].mean() * 100,
        df.loc[df["Age"] <= 25, "Attrition"].mean() * 100,
        df.loc[df["YearsAtCompany"] <= 2, "Attrition"].mean() * 100,
        df.loc[df["JobRole"] == "Sales Representative", "Attrition"].mean() * 100,
        df.loc[df["JobRole"] == "Laboratory Technician", "Attrition"].mean() * 100
    ]
})

hotspots
```

# Key Findings

1. The overall attrition rate is 16.12%.

2. Sales has the highest attrition rate among the three major departments.

3. Sales Representatives have the highest observed attrition rate among job roles.

4. Employees aged 25 or below show substantially higher observed attrition.

5. Employees with 0–2 years at the company have considerably higher attrition than longer-tenure groups.

6. Employees working overtime have substantially higher observed attrition than employees without overtime.

7. Employees who travel frequently have higher observed attrition than non-travel employees.

8. Employees living farther from the workplace show higher observed attrition in this dataset.

9. These findings represent descriptive associations and should not be interpreted as proof of causation.

10. The analysis should be used to identify areas for further HR investigation rather than to label individual employees as high-risk.
